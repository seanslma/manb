## read excel using camline


```python
from pathlib import Path

xlfl = Path.home().joinpath('dat/test/cola.xlsx')
wsnm = ['COCA COLA CO']
```

```python
import pandas as pd

dp = pd.read_excel(
    xlfl,
    sheet_name=wsnm[0],
    skiprows=3,
    skipfooter=0,
    engine='calamine',
    dtype_backend='pyarrow',
)
dp[:2]
```

```python
import polars as pl

dl = pl.read_excel(
    xlfl,
    sheet_name=wsnm[0],
    read_options={'header_row': 3},
    engine='calamine',
)

dl[:2]
```
